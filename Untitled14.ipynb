```python
import zipfile

zip_path = "/content/Underwater-Image-Data-set-main.zip"
extract_path = "/content/underwater_dataset"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_path)
```

```python
# Install 7zip
!apt-get -qq install p7zip-full

#Dataset folder
dataset = "/content/underwater_dataset/Underwater-Image-Data-set-main"

#Folder where images will be extracted
output = "/content/images"

# Extract all .7z files
!mkdir -p /content/images
!7z x "$dataset/*.7z" -o"$output" -y
```

```python
#Count PNG images
import os

images = []
for root, dirs, files in os.walk(output):
    for file in files:
        if file.lower().endswith((".png", ".jpg", ".jpeg")):
            images.append(os.path.join(root, file))

print("Total images:", len(images))

#Show first 10 image paths
print("\nFirst 10 images:")
for img in images[:10]:
    print(img)
```

```python
#sample images
from PIL import Image
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

for i, img_path in enumerate(images[:6]):
    img = Image.open(img_path)

    plt.subplot(2, 3, i + 1)
    plt.imshow(img)
    plt.axis("off")
    plt.title(os.path.basename(img_path))

plt.show()
```

```python
#Image Dimensions
processed_image_count = 0
MAX_IMAGES_TO_SHOW = 10 # Limit to the first 10 images for display

for root, dirs, files in os.walk(dataset):
    for file in files:
        if file.lower().endswith(".png"): # Process only PNG files, case-insensitive
            if processed_image_count >= MAX_IMAGES_TO_SHOW:
                break # Exit inner loop if max images processed

            path = os.path.join(root, file)
            try:
                img = Image.open(path)
                print(f"{file} -> {img.size}")
                processed_image_count += 1
            except Exception as e:
                # Catch specific exceptions for better error handling, or log them.
                # For now, using 'pass' as in the original intent, but consider logging.
                pass
    # If the inner loop broke because MAX_IMAGES_TO_SHOW was reached, also break the outer loop
    if processed_image_count >= MAX_IMAGES_TO_SHOW:
        break
```
